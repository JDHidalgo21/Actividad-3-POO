Ejercicio 8.2: Notas

```python
import math
import tkinter as tk
from tkinter import messagebox


class Notas:
    """
    Clase que representa la lógica de negocio para gestionar la lista de notas de un estudiante
    y calcular sus estadísticas asociadas.
    """
    def __init__(self):
        self.lista_notas = []

    def agregar_nota(self, nota: float) -> None:
        """Agrega una nota a la lista tras validar que esté en el rango permitido [0.0, 5.0]."""
        if 0.0 <= nota <= 5.0:
            self.lista_notas.append(nota)
        else:
            raise ValueError("La nota debe estar entre 0.0 y 5.0")

    def calcular_promedio(self) -> float:
        """Calcula el promedio aritmético de las notas ingresadas."""
        if not self.lista_notas:
            return 0.0
        return sum(self.lista_notas) / len(self.lista_notas)

    def calcular_desviacion_estandar(self) -> float:
        """Calcula la desviación estándar poblacional de las notas ingresadas."""
        if not self.lista_notas:
            return 0.0
        promedio = self.calcular_promedio()
        suma_cuadrados = sum((x - promedio) ** 2 for x in self.lista_notas)
        varianza = suma_cuadrados / len(self.lista_notas)
        return math.sqrt(varianza)

    def calcular_mayor(self) -> float:
        """Obtiene la nota mayor de la lista."""
        if not self.lista_notas:
            return 0.0
        return max(self.lista_notas)

    def calcular_menor(self) -> float:
        """Obtiene la nota menor de la lista."""
        if not self.lista_notas:
            return 0.0
        return min(self.lista_notas)

    def limpiar(self) -> None:
        """Vacía la lista de notas."""
        self.lista_notas.clear()


class VentanaNotas(tk.Tk):
    """
    Clase que construye la Interfaz Gráfica de Usuario (GUI) utilizando Tkinter.
    """
    def __init__(self):
        super().__init__()
        self.notas = Notas()

        self.title("Notas")
        self.geometry("280x380")
        self.resizable(False, False)

        self._crear_componentes()

    def _crear_componentes(self):
        # Campos de texto para las 5 notas
        self.entries_notas = []
        for i in range(5):
            lbl = tk.Label(self, text=f"Nota {i + 1}:")
            lbl.place(x=20, y=20 + (i * 30), width=60, height=23)

            entry = tk.Entry(self)
            entry.place(x=90, y=20 + (i * 30), width=150, height=23)
            self.entries_notas.append(entry)

        # Botón Calcular
        self.btn_calcular = tk.Button(self, text="Calcular", command=self._calcular_estadisticas)
        self.btn_calcular.place(x=40, y=180, width=90, height=25)

        # Botón Limpiar
        self.btn_limpiar = tk.Button(self, text="Limpiar", command=self._limpiar_campos)
        self.btn_limpiar.place(x=150, y=180, width=90, height=25)

        # Etiquetas de resultados
        self.lbl_promedio = tk.Label(self, text="Promedio = ", anchor="w")
        self.lbl_promedio.place(x=20, y=220, width=240, height=23)

        self.lbl_desviacion = tk.Label(self, text="Desviación estándar = ", anchor="w")
        self.lbl_desviacion.place(x=20, y=250, width=240, height=23)

        self.lbl_mayor = tk.Label(self, text="Valor mayor = ", anchor="w")
        self.lbl_mayor.place(x=20, y=280, width=240, height=23)

        self.lbl_menor = tk.Label(self, text="Valor menor = ", anchor="w")
        self.lbl_menor.place(x=20, y=310, width=240, height=23)

    def _calcular_estadisticas(self):
        self.notas.limpiar()
        try:
            for entry in self.entries_notas:
                texto = entry.get().strip()
                if not texto:
                    raise ValueError("Campo nulo o vacío")
                val = float(texto)
                self.notas.agregar_nota(val)

            # Actualización de etiquetas
            self.lbl_promedio.config(text=f"Promedio = {self.notas.calcular_promedio():.2f}")
            self.lbl_desviacion.config(
                text=f"Desviación estándar = {self.notas.calcular_desviacion_estandar():.2f}"
            )
            self.lbl_mayor.config(text=f"Valor mayor = {self.notas.calcular_mayor():.2f}")
            self.lbl_menor.config(text=f"Valor menor = {self.notas.calcular_menor():.2f}")

        except ValueError:
            messagebox.showerror("Error", "Campo nulo o error en formato de número")

    def _limpiar_campos(self):
        for entry in self.entries_notas:
            entry.delete(0, tk.END)
        self.notas.limpiar()

        self.lbl_promedio.config(text="Promedio = ")
        self.lbl_desviacion.config(text="Desviación estándar = ")
        self.lbl_mayor.config(text="Valor mayor = ")
        self.lbl_menor.config(text="Valor menor = ")


if __name__ == "__main__":
    app = VentanaNotas()
    app.mainloop()
```

Ejercicio 8.3: Figuras Geométricas (Cilindro, Esfera y Pirámide)

```python
import math
import tkinter as tk
from tkinter import messagebox

# Clases de Lógica

class FiguraGeometrica:
    def __init__(self):
        self.volumen = 0.0
        self.superficie = 0.0

    def calcular_volumen(self):
        pass

    def calcular_superficie(self):
        pass


class Cilindro(FiguraGeometrica):
    def __init__(self, radio: float, altura: float):
        super().__init__()
        self.radio = radio
        self.altura = altura
        self.calcular_volumen()
        self.calcular_superficie()

    def calcular_volumen(self):
        self.volumen = math.pi * (self.radio ** 2) * self.altura
        return self.volumen

    def calcular_superficie(self):
        area_lateral = 2 * math.pi * self.radio * self.altura
        area_base = math.pi * (self.radio ** 2)
        self.superficie = area_lateral + (2 * area_base)
        return self.superficie


class Esfera(FiguraGeometrica):
    def __init__(self, radio: float):
        super().__init__()
        self.radio = radio
        self.calcular_volumen()
        self.calcular_superficie()

    def calcular_volumen(self):
        self.volumen = (4 / 3) * math.pi * (self.radio ** 3)
        return self.volumen

    def calcular_superficie(self):
        self.superficie = 4 * math.pi * (self.radio ** 2)
        return self.superficie


class Piramide(FiguraGeometrica):
    def __init__(self, base: float, altura: float, apotema: float):
        super().__init__()
        self.base = base
        self.altura = altura
        self.apotema = apotema
        self.calcular_volumen()
        self.calcular_superficie()

    def calcular_volumen(self):
        self.volumen = ((self.base ** 2) * self.altura) / 3.0
        return self.volumen

    def calcular_superficie(self):
        area_base = self.base ** 2
        area_triangulos = 2 * self.base * self.apotema
        self.superficie = area_base + area_triangulos
        return self.superficie

#Interfaz Gráfica

class VentanaCilindro(tk.Toplevel):
    def __init__(self, parent):
        super().__init__(parent)
        self.title("Cilindro")
        self.geometry("280x210")
        self.resizable(False, False)

        tk.Label(self, text="Radio (cm):").grid(row=0, column=0, padx=10, pady=10, sticky="e")
        self.txt_radio = tk.Entry(self, width=15)
        self.txt_radio.grid(row=0, column=1, padx=10, pady=10)

        tk.Label(self, text="Altura (cm):").grid(row=1, column=0, padx=10, pady=10, sticky="e")
        self.txt_altura = tk.Entry(self, width=15)
        self.txt_altura.grid(row=1, column=1, padx=10, pady=10)

        tk.Button(self, text="Calcular", command=self.calcular).grid(row=2, column=0, columnspan=2, pady=10)

        self.lbl_volumen = tk.Label(self, text="Volumen (cm³): ")
        self.lbl_volumen.grid(row=3, column=0, columnspan=2, sticky="w", padx=20)

        self.lbl_superficie = tk.Label(self, text="Superficie (cm²): ")
        self.lbl_superficie.grid(row=4, column=0, columnspan=2, sticky="w", padx=20)

    def calcular(self):
        try:
            radio = float(self.txt_radio.get())
            altura = float(self.txt_altura.get())

            if radio <= 0 or altura <= 0:
                messagebox.showerror("Error", "Los valores deben ser positivos.")
                return

            cilindro = Cilindro(radio, altura)
            self.lbl_volumen.config(text=f"Volumen (cm³): {cilindro.volumen:.2f}")
            self.lbl_superficie.config(text=f"Superficie (cm²): {cilindro.superficie:.2f}")
        except ValueError:
            messagebox.showerror("Error de formato", "Por favor ingrese valores numéricos válidos.")


class VentanaEsfera(tk.Toplevel):
    def __init__(self, parent):
        super().__init__(parent)
        self.title("Esfera")
        self.geometry("280x180")
        self.resizable(False, False)

        tk.Label(self, text="Radio (cm):").grid(row=0, column=0, padx=10, pady=10, sticky="e")
        self.txt_radio = tk.Entry(self, width=15)
        self.txt_radio.grid(row=0, column=1, padx=10, pady=10)

        tk.Button(self, text="Calcular", command=self.calcular).grid(row=1, column=0, columnspan=2, pady=10)

        self.lbl_volumen = tk.Label(self, text="Volumen (cm³): ")
        self.lbl_volumen.grid(row=2, column=0, columnspan=2, sticky="w", padx=20)

        self.lbl_superficie = tk.Label(self, text="Superficie (cm²): ")
        self.lbl_superficie.grid(row=3, column=0, columnspan=2, sticky="w", padx=20)

    def calcular(self):
        try:
            radio = float(self.txt_radio.get())

            if radio <= 0:
                messagebox.showerror("Error", "El radio debe ser positivo.")
                return

            esfera = Esfera(radio)
            self.lbl_volumen.config(text=f"Volumen (cm³): {esfera.volumen:.2f}")
            self.lbl_superficie.config(text=f"Superficie (cm²): {esfera.superficie:.2f}")
        except ValueError:
            messagebox.showerror("Error de formato", "Por favor ingrese un valor numérico válido.")


class VentanaPiramide(tk.Toplevel):
    def __init__(self, parent):
        super().__init__(parent)
        self.title("Pirámide")
        self.geometry("280x240")
        self.resizable(False, False)

        tk.Label(self, text="Base (cm):").grid(row=0, column=0, padx=10, pady=5, sticky="e")
        self.txt_base = tk.Entry(self, width=15)
        self.txt_base.grid(row=0, column=1, padx=10, pady=5)

        tk.Label(self, text="Altura (cm):").grid(row=1, column=0, padx=10, pady=5, sticky="e")
        self.txt_altura = tk.Entry(self, width=15)
        self.txt_altura.grid(row=1, column=1, padx=10, pady=5)

        tk.Label(self, text="Apotema (cm):").grid(row=2, column=0, padx=10, pady=5, sticky="e")
        self.txt_apotema = tk.Entry(self, width=15)
        self.txt_apotema.grid(row=2, column=1, padx=10, pady=5)

        tk.Button(self, text="Calcular", command=self.calcular).grid(row=3, column=0, columnspan=2, pady=10)

        self.lbl_volumen = tk.Label(self, text="Volumen (cm³): ")
        self.lbl_volumen.grid(row=4, column=0, columnspan=2, sticky="w", padx=20)

        self.lbl_superficie = tk.Label(self, text="Superficie (cm²): ")
        self.lbl_superficie.grid(row=5, column=0, columnspan=2, sticky="w", padx=20)

    def calcular(self):
        try:
            base = float(self.txt_base.get())
            altura = float(self.txt_altura.get())
            apotema = float(self.txt_apotema.get())

            if base <= 0 or altura <= 0 or apotema <= 0:
                messagebox.showerror("Error", "Todos los valores deben ser positivos.")
                return

            piramide = Piramide(base, altura, apotema)
            self.lbl_volumen.config(text=f"Volumen (cm³): {piramide.volumen:.2f}")
            self.lbl_superficie.config(text=f"Superficie (cm²): {piramide.superficie:.2f}")
        except ValueError:
            messagebox.showerror("Error de formato", "Por favor ingrese valores numéricos válidos.")


class VentanaPrincipal(tk.Tk):
    def __init__(self):
        super().__init__()
        self.title("Figuras Geométricas")
        self.geometry("260x180")
        self.resizable(False, False)

        tk.Label(self, text="Seleccione una figura:", font=("Arial", 10, "bold")).pack(pady=15)
        tk.Button(self, text="Cilindro", command=lambda: VentanaCilindro(self)).pack(fill="x", padx=30, pady=3)
        tk.Button(self, text="Esfera", command=lambda: VentanaEsfera(self)).pack(fill="x", padx=30, pady=3)
        tk.Button(self, text="Pirámide", command=lambda: VentanaPiramide(self)).pack(fill="x", padx=30, pady=3)


# Bloque de ejecución

if __name__ == "__main__":
    try:
        app = VentanaPrincipal()
        app.mainloop()
    except Exception as e:
        print("--- Entorno sin servidor gráfico detectado (ej. Google Colab) ---")
        print("--- Ejecutando pruebas unitarias por consola ---\n")

        c = Cilindro(radio=5.0, altura=10.0)
        print(f"Cilindro (r=5, h=10) -> Volumen: {c.volumen:.2f} cm³, Superficie: {c.superficie:.2f} cm²")

        e = Esfera(radio=4.0)
        print(f"Esfera (r=4)        -> Volumen: {e.volumen:.2f} cm³, Superficie: {e.superficie:.2f} cm²")

        p = Piramide(base=6.0, altura=8.0, apotema=8.5)
        print(f"Pirámide (b=6, h=8)  -> Volumen: {p.volumen:.2f} cm³, Superficie: {p.superficie:.2f} cm²")
```
